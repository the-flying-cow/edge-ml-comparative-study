# Imports

```python
import numpy as np
import pandas as pd

from sklearn.datasets import make_classification

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, export_text

from sklearn.metrics import classification_report
```

Creating a synthetic dataset using the 'make_classification' function from the sklearn library.

```python
np.random.seed(7)
```

```python
features, target= make_classification(n_samples= 1000,
                                      n_features= 5,
                                      n_informative= 2,
                                      n_classes= 2,
                                      random_state= 7,
                                      weights= [0.2, 0.8])

data= pd.DataFrame(features, columns= ['Feature1', 'Feature2', 'Feature3', 'Feature4', 'Feature5'])
target= pd.Series(target, name= 'Label')
```

Use StandardScaler for scaling features to similar range.

```python
scaler= StandardScaler()
scaled_data= scaler.fit_transform(data)
scaled_data= pd.DataFrame(scaled_data, columns= data.columns)
```

```python
scaled_data.head(3)
```

```python
X_train, X_test, y_train, y_test= train_test_split(scaled_data, target, test_size= 0.2, stratify= target, random_state= 7)
```

```python
models= {
    "LR": LogisticRegression(class_weight= 'balanced', random_state= 7),
    "DT": DecisionTreeClassifier(max_depth=3, class_weight= 'balanced',random_state= 7)
}
```

```python
for name,model in models.items():
  print(name)

  model.fit(X_train, y_train)
  preds= model.predict(X_test)

  print(classification_report(y_test, preds))
```

```python
print(f"Mean:{scaler.mean_}\nScale:{scaler.scale_}")
```

```python
print(f"Coefficients:{models["LR"].coef_}\nIntercept:{models["LR"].intercept_}")
```

```python
print(f"{export_text(models["DT"])}")
```
